```python
import pandas as pd
import glob
import os

# Lấy danh sách 24 file Excel
files = glob.glob("data_excel/*.xlsx")

print("Tổng số file:", len(files))

# Danh sách lưu dữ liệu
all_data = []

# Đọc từng file
for file in sorted(files):

    df = pd.read_excel(file)

    # Lấy mã cổ phiếu từ tên file
    symbol = os.path.basename(file).split("_")[0]

    # Gắn mã cổ phiếu
    df["Symbol"] = symbol

    # Gắn tên file để truy xuất nguồn gốc
    df["Source_File"] = os.path.basename(file)

    all_data.append(df)

print("Đã đọc xong toàn bộ file!")
```

```python
# Gộp 24 file thành một DataFrame
raw = pd.concat(all_data, ignore_index=True)

print("Kích thước dữ liệu:", raw.shape)

print("\n5 dòng đầu tiên:")
display(raw.head())

print("\nTên các cột:")
print(raw.columns.tolist())
```

```python
# Chuyển ngày thành kiểu datetime
raw["Ngày"] = pd.to_datetime(
    raw["Ngày"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Đổi tên cột
raw = raw.rename(columns={
    "Mã": "Original_Symbol",
    "Symbol": "Symbol",
    "Ngày": "Date",
    "Giá đóng cửa": "Close",
    "Giá điều chỉnh": "Adj_Close",
    "Thay đổi": "Change",
    "Khối lượng khớp lệnh": "Volume",
    "Giá trị khớp lệnh": "Trading_Value",
    "Khối lượng thỏa thuận": "Negotiated_Volume",
    "Giá trị thỏa thuận": "Negotiated_Value",
    "Giá mở cửa": "Open",
    "Giá cao nhất": "High",
    "Giá thấp nhất": "Low"
})

# Sắp xếp theo mã và ngày tăng dần
raw = raw.sort_values(
    by=["Symbol", "Date"]
).reset_index(drop=True)

print("Đã chuẩn hóa dữ liệu!")
display(raw.head())
```

```python
# Sao chép dữ liệu sạch
feature_df = clean.copy()

# Chuyển ngày sang datetime
feature_df["Date"] = pd.to_datetime(feature_df["Date"])

# Sắp xếp theo mã và ngày
feature_df = feature_df.sort_values(
    ["Symbol", "Date"]
).reset_index(drop=True)

# Tạo Return
feature_df["Return"] = (
    feature_df.groupby("Symbol")["Adj_Close"]
    .pct_change()
)

# Tạo MA5
feature_df["MA5"] = (
    feature_df.groupby("Symbol")["Adj_Close"]
    .transform(lambda x: x.rolling(5).mean())
)

# Tạo MA20
feature_df["MA20"] = (
    feature_df.groupby("Symbol")["Adj_Close"]
    .transform(lambda x: x.rolling(20).mean())
)

# Tạo Lag 1
feature_df["Lag_1"] = (
    feature_df.groupby("Symbol")["Adj_Close"].shift(1)
)

# Tạo Lag 2
feature_df["Lag_2"] = (
    feature_df.groupby("Symbol")["Adj_Close"].shift(2)
)

display(feature_df.head(25))
```

```python
# Lấy danh sách ngày duy nhất
dates = sorted(feature_df["Date"].unique())

# Chọn mốc chia 80/20
split_index = int(len(dates) * 0.8)
split_date = dates[split_index]

# Chia dữ liệu
train = feature_df[feature_df["Date"] < split_date].copy()
test = feature_df[feature_df["Date"] >= split_date].copy()

print("Ngày bắt đầu Train:", train["Date"].min())
print("Ngày kết thúc Train:", train["Date"].max())

print("\nNgày bắt đầu Test:", test["Date"].min())
print("Ngày kết thúc Test:", test["Date"].max())

print("\nKích thước Train:", train.shape)
print("Kích thước Test:", test.shape)
```

```python
features = [
    "Return",
    "MA5",
    "MA20",
    "Lag_1",
    "Lag_2"
]

X_train = train[features]
y_train = train["Target"]

X_test = test[features]
y_test = test["Target"]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Chuẩn hóa dữ liệu thành công!")
```

```python

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix
)

# 1. Sắp xếp dữ liệu theo thời gian
train_sorted = train.sort_values("Date").copy()
test_sorted = test.sort_values("Date").copy()

X = train_sorted[features]
y = train_sorted["Target"]

X_test_new = test_sorted[features]
y_test_new = test_sorted["Target"]

# 2. Chia Train thành Train nhỏ và Validation theo thời gian
split = int(len(X) * 0.8)

X_subtrain = X.iloc[:split]
X_val = X.iloc[split:]

y_subtrain = y.iloc[:split]
y_val = y.iloc[split:]

# 3. Thử nhiều tham số khác nhau
best_f1 = -1
best_params = None

for c in [0.01, 0.1, 1, 10, 100]:
    for weight in [None, "balanced"]:

        scaler_temp = StandardScaler()

        X_sub_scaled = scaler_temp.fit_transform(X_subtrain)
        X_val_scaled = scaler_temp.transform(X_val)

        temp_model = LogisticRegression(
            C=c,
            class_weight=weight,
            max_iter=5000
        )

        temp_model.fit(X_sub_scaled, y_subtrain)

        probabilities = temp_model.predict_proba(
            X_val_scaled
        )[:, 1]

        # Thử nhiều ngưỡng dự đoán
        for threshold in np.arange(0.20, 0.81, 0.02):

            predictions = (
                probabilities >= threshold
            ).astype(int)

            f1 = f1_score(
                y_val,
                predictions,
                zero_division=0
            )

            if f1 > best_f1:
                best_f1 = f1

                best_params = {
                    "C": c,
                    "class_weight": weight,
                    "threshold": threshold
                }

# 4. Hiển thị tham số tốt nhất
print("===== THAM SỐ TỐI ƯU =====")
print(best_params)
print("Validation F1-score:", best_f1)

# 5. Huấn luyện lại trên toàn bộ Train
scaler = StandardScaler()

X_train_new = scaler.fit_transform(X)
X_test_scaled_new = scaler.transform(X_test_new)

model = LogisticRegression(
    C=best_params["C"],
    class_weight=best_params["class_weight"],
    max_iter=5000
)

model.fit(X_train_new, y)

# 6. Dự đoán trên Test
proba = model.predict_proba(X_test_scaled_new)[:, 1]

# So sánh ngưỡng mặc định và ngưỡng tối ưu
y_pred_default = (proba >= 0.5).astype(int)

y_pred = (
    proba >= best_params["threshold"]
).astype(int)

# 7. Đánh giá mô hình
print("\n===== NGƯỠNG 0.5 =====")
print("Accuracy:", accuracy_score(y_test_new, y_pred_default))
print("F1-score:", f1_score(y_test_new, y_pred_default))

print("\n===== NGƯỠNG TỐI ƯU =====")
print("Accuracy:", accuracy_score(y_test_new, y_pred))
print("Precision:", precision_score(y_test_new, y_pred, zero_division=0))
print("Recall:", recall_score(y_test_new, y_pred, zero_division=0))
print("F1-score:", f1_score(y_test_new, y_pred, zero_division=0))
print("Balanced Accuracy:", balanced_accuracy_score(y_test_new, y_pred))

print("\nClassification Report:")
print(classification_report(y_test_new, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_new, y_pred))
```

```python

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    classification_report
)

# 1. Sắp xếp dữ liệu theo thời gian
data_train = train.copy()
data_test = test.copy()

data_train["Date"] = pd.to_datetime(data_train["Date"])
data_test["Date"] = pd.to_datetime(data_test["Date"])

data_train = data_train.sort_values("Date")

# 2. Chia Train thành Train nhỏ và Validation
dates = sorted(data_train["Date"].unique())
split_date = dates[int(len(dates) * 0.8)]

subtrain = data_train[data_train["Date"] < split_date]
val = data_train[data_train["Date"] >= split_date]

X_sub = subtrain[features]
y_sub = subtrain["Target"]

X_val = val[features]
y_val = val["Target"]

X_test_final = data_test[features]
y_test_final = data_test["Target"]

# 3. Chuẩn hóa dữ liệu
scaler_final = StandardScaler()

X_sub_scaled = scaler_final.fit_transform(X_sub)
X_val_scaled = scaler_final.transform(X_val)
X_test_final_scaled = scaler_final.transform(X_test_final)

# 4. Tìm tham số tốt nhất
results = []

for c in [0.001, 0.01, 0.1, 1, 10, 100]:
    for weight in [None, "balanced"]:

        temp_model = LogisticRegression(
            C=c,
            class_weight=weight,
            max_iter=5000
        )

        temp_model.fit(X_sub_scaled, y_sub)

        probabilities = temp_model.predict_proba(
            X_val_scaled
        )[:, 1]

        for threshold in np.arange(0.20, 0.81, 0.02):

            pred = (probabilities >= threshold).astype(int)

            # Không nhận mô hình chỉ dự đoán một lớp
            if len(np.unique(pred)) < 2:
                continue

            results.append({
                "C": c,
                "Weight": weight,
                "Threshold": threshold,
                "Accuracy": accuracy_score(y_val, pred),
                "Balanced_Accuracy": balanced_accuracy_score(y_val, pred),
                "Precision": precision_score(y_val, pred, zero_division=0),
                "Recall": recall_score(y_val, pred, zero_division=0),
                "F1": f1_score(y_val, pred, zero_division=0)
            })

# 5. Xếp hạng kết quả Validation
result_df = pd.DataFrame(results)

result_df = result_df.sort_values(
    ["Balanced_Accuracy", "F1"],
    ascending=False
)

print("===== TOP 10 MÔ HÌNH VALIDATION =====")
display(result_df.head(10))

# 6. Chọn mô hình tốt nhất
best = result_df.iloc[0]

print("\nTHAM SỐ TỐT NHẤT:")
print(best)

# 7. Đánh giá trên Test bằng mô hình đã chọn
final_model = LogisticRegression(
    C=float(best["C"]),
    class_weight=best["Weight"],
    max_iter=5000
)

final_model.fit(X_sub_scaled, y_sub)

test_prob = final_model.predict_proba(
    X_test_final_scaled
)[:, 1]

final_pred = (
    test_prob >= float(best["Threshold"])
).astype(int)

print("\n===== KẾT QUẢ TEST =====")
print("Accuracy:", accuracy_score(y_test_final, final_pred))
print("Balanced Accuracy:", balanced_accuracy_score(y_test_final, final_pred))
print("Precision:", precision_score(y_test_final, final_pred, zero_division=0))
print("Recall:", recall_score(y_test_final, final_pred, zero_division=0))
print("F1-score:", f1_score(y_test_final, final_pred, zero_division=0))

print("\nClassification Report:")
print(classification_report(y_test_final, final_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_final, final_pred))
```
